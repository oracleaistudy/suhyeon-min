**09 텍스트를 위한 인공 신경망**
====

**09-1 순차 데이터와 순환 신경망**
----

**순차 데이터 sequential data** : 순서에 의미가 있는 데이터     
- 텍스트 ex) I am a boy      
- 시계열 ex) 1일 15도, 2일 17도      

**순환 신경망 recurrent neutral network, RNN**   
이전 데이터의 처리 결과가 순환됨 - 은닉층     

- 타임스텝 timestep : 한 뉴런에 샘플을 처리하는 한 단계     
-> 오래될수록 순환되는 정보는 희미해진다.    

- 셀 cell : 순환층      

- 은닉 상태 hidden state : 순환 신경망에서 활성화 함수를 통과한 출력    
-> 다음 타임 스텝에 재사용      

- 활성화 함수 : 탄젠트 hyperbolic tangent (**tanh**), -1 ~ 1의 범위를 가짐     

- 가중치     
타임스텝1 : Wx(입력에 곱해지는 가중치) * h0 (모두 0으로 사용) -> h1 (출력)       
타임스텝2 : Wx(입력에 곱해지는 가중치) * h1 * Wh(타임스텝의 은닉 상태에 곱해지는 가중치) -> h2 (출력)         
타임스텝3 : Wx(입력에 곱해지는 가중치) * h2 * Wh -> h3 (출력)      

가중치 개수 = 모델 파라미터 수 = Wx + Wh + b        
ex) 입력층 4 / 순환층 3 일때, Wx = 4 * 3 / Wh = 3 * 3 / b = 3     
즉 24개     

- 순환층의 입력과 출력     
입력 : 2차원 배열 (타임스텝 = 시퀀스 길이, 단어 표현)     
출력 : 1차원 배열 (뉴련 개수)          

- 출력층의 구성    
밀집층 : 클래스 분류 (다중 분류 : 소프트맥스 활성화 함수, 이진 분류 : 시그모이드 활성화 함수)      
마지막 셀의 출력이 1차원이라서 Flatten 클래스로 펼칠 필요가 없음
